Exercise 3 — Self Reflection Prompt

Goal:
Have the AI critique and improve its own summary.

```python
# Original summary

summary_before = """
Artificial intelligence is becoming very popular and many companies use it to improve productivity.
"""

print("Original Summary:")
print(summary_before)
```

```python
# Self-reflection prompt

reflection_prompt = """
Review the following summary.

Criteria:
- Improve clarity
- Make it concise
- Limit to 2 sentences
- Focus on business impact
- Professional tone

Then produce an improved summary.
"""

print(reflection_prompt)
```

```python
# Improved summary

summary_after = """
Artificial intelligence is widely adopted by organizations to improve efficiency and automate tasks.
Businesses use AI to enhance decision-making, productivity, and customer service.
"""

print("\nImproved Summary:")
print(summary_after)
```
