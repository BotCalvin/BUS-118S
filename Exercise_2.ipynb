Exercise 2 — ReACT Code Generation

Goal:
Generate Python code to calculate the average of a list of numbers.

ReACT Process:
1. Reason about the task
2. Generate code
3. Run code
4. Observe output
5. Fix if needed

```python
# REASON
# We need to compute the average of a list.
# Steps:
# 1. Sum numbers
# 2. Divide by length

numbers = [10, 20, 30, 40, 50]
```

```python
# ACT — generate and run code

def calculate_average(nums):
    return sum(nums) / len(nums)

result = calculate_average(numbers)

print("Average:", result)
```

```python
# OBSERVE
# Check if result is correct

expected = 30

if result == expected:
    print("Output is correct")
else:
    print("Error detected")
```

```python
# FIX (example edge case)

def safe_average(nums):
    if len(nums) == 0:
        return "List is empty"
    return sum(nums) / len(nums)

print("Safe Average:", safe_average(numbers))
```
